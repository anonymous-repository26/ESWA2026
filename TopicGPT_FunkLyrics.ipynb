```python
import pandas as pd
file_path = '/content/drive/MyDrive/UFRGS/ESWA/funk_lyrics_estrofes_1000.xlsx'

df = pd.read_excel(file_path)
print(f"Shape: {df.shape}")
print(f"Colunas: {df.columns.tolist()}")
display(df.head())
```

```python
# Converte para lista e remove documentos nulos/vazios
corpus_raw = df['text'].tolist()
corpus = [str(doc).strip() for doc in corpus_raw if pd.notna(doc) and str(doc).strip() != '']

print(f"Total de documentos após limpeza: {len(corpus)}")
print(f"\nExemplo de documento:\n{corpus[0][:300]}")
```

```python


# =========================
# INICIALIZAR MODELO
# =========================

tm = TopicGPT(
    openai_api_key=OPENAI_API_KEY,
    n_topics=10
)

# =========================
# TREINAR MODELO
# =========================

tm.fit(corpus)
```

```python

# =========================
# LISTAR TÓPICOS
# =========================

print(tm.topic_lis)
```

```python
# =========================
# IMPRIMIR TÓPICOS
# =========================

tm.print_topics()
```

```python
# =========================
# VISUALIZAÇÃO
# =========================

tm.visualize_clusters()
```
